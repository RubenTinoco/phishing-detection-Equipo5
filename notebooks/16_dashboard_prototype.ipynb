# Sprint 5 - Dashboard Prototype

Este notebook implementa PB-17 a nivel de prototipo: define el wireframe, valida los KPIs principales y documenta como ejecutar `dashboard/app.py`.

## 1. Objetivo del dashboard

El dashboard debe comunicar resultados del modelo final a stakeholders no tecnicos: metricas, matriz de confusion, simulador de predicciones y explicabilidad.

```python
from pathlib import Path
import pandas as pd

ROOT = Path('..')
metrics = pd.read_csv(ROOT / 'models' / 'final_comparison.csv', index_col=0)
summary = pd.read_csv(ROOT / 'reports' / 'business_value_summary.csv')
metrics
```

## 2. KPIs del primer viewport

Los KPIs principales del dashboard son F1, AUC-ROC, Recall, Precision y valor anual estimado.

```python
final_metrics = metrics.loc['Modelo Final (Sprint 4)'].to_dict()
business_summary = dict(zip(summary['item'], summary['value']))

kpis = pd.DataFrame([
    {'KPI': 'F1 Score', 'Valor': final_metrics['F1']},
    {'KPI': 'AUC-ROC', 'Valor': final_metrics['AUC-ROC']},
    {'KPI': 'Recall', 'Valor': final_metrics['Recall']},
    {'KPI': 'Precision', 'Valor': final_metrics['Precision']},
    {'KPI': 'Valor anual estimado USD', 'Valor': business_summary.get('annual_value_reference_usd')},
])
kpis
```

## 3. Wireframe propuesto

1. Sidebar: umbral y supuestos de negocio.
2. Encabezado: KPIs principales.
3. Tab Resumen: matriz de confusion y matriz costo-beneficio.
4. Tab Predicciones: carga de CSV y tabla con probabilidad de phishing.
5. Tab Explicabilidad: importancia global de variables y detalle por instancia.

```python
dashboard_path = ROOT / 'dashboard' / 'app.py'
print(f'Dashboard Streamlit: {dashboard_path}')
print('Comando de ejecucion: streamlit run dashboard/app.py')
```
